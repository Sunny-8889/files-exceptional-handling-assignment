Q1. What is the difference between interpreted and compiled languages?

Ans - Compiled languages are typically translated into machine code or another lower-level form before execution. This can make execution faster, although a separate compilation step is usually required. Interpreted languages are commonly executed by an interpreter at runtime, often statement by statement or through an intermediate representation. Python is often called interpreted, but its implementation usually compiles source code to bytecode first and then runs that bytecode in the Python virtual machine. The exact distinction depends on the language implementation.


Q2. What is exception handling in Python?

Ans - Exception handling is the process of responding to runtime errors so a program can recover gracefully or stop in a controlled way. Python uses try, except, else, and finally blocks to detect and manage exceptions. For example, a program can catch ValueError when converting invalid input to a number instead of crashing unexpectedly.


Q3. What is the purpose of the finally block in exception handling?

Ans - The finally block contains cleanup code that should run whether an exception occurs or not. It is commonly used to release resources, close files, or clean up connections. A finally block normally runs after the try and except/else sections, even when an exception is raised; process termination or other extreme circumstances can prevent it from running.


Q4. What is logging in Python?

Ans - Logging is the practice of recording messages about a program's operation, such as progress, warnings, errors, and debugging details. Python provides the built-in logging module, which can send messages to the console, a file, or other handlers. Log records can include timestamps, severity levels, and contextual information.


Q5. What is the significance of the __del__ method in Python?

Ans - The __del__ method is a finalizer that Python may call when an object is being finalized. It can be used for cleanup, but its timing is not guaranteed and it may behave unexpectedly with reference cycles or interpreter shutdown. It is generally better to use context managers (with) or explicit cleanup methods for resources such as files and sockets.


Q6. What is the difference between import and from ... import in Python?

Ans - import module imports the module and normally requires its name to access members, for example, import math followed by math.sqrt(9). from module import name imports a selected member directly into the current namespace, for example, from math import sqrt followed by sqrt(9). The second form can be more concise but may cause naming conflicts and makes the origin of a name less obvious.


Q7. How can you handle multiple exceptions in Python?

Ans - Use separate except blocks when different exceptions need different responses, or group exception classes in a tuple when they should be handled the same way. For example: except (ValueError, TypeError) as error: ... Catch specific exceptions where possible and avoid a broad except Exception unless there is a good reason.


Q8. What is the purpose of the with statement when handling files in Python?

Ans - The with statement uses a context manager to manage a resource. When a file is opened with with open(...), Python automatically closes the file when the block finishes, including when an exception occurs. This makes file handling safer and avoids relying on manual close() calls.


Q9. What is the difference between multithreading and multiprocessing?

Ans - Multithreading runs multiple threads within one process. Threads share the process's memory, so sharing data can be convenient, but synchronization may be needed. In standard CPython, the Global Interpreter Lock (GIL) limits simultaneous execution of Python bytecode by threads, so threads are often most useful for I/O-bound tasks. Multiprocessing uses separate processes with separate memory spaces and can use multiple CPU cores for CPU-bound work, but it usually has greater startup and data-sharing overhead.


Q10. What are the advantages of using logging in a program?

Ans - Logging helps developers diagnose errors, monitor program behavior, and understand what happened before a failure. It supports severity levels, timestamps, output to files or other destinations, configurable message formats, and filtering. Unlike scattered print statements, logging can be adjusted for development or production without changing the program's core logic.


Q11. What is memory management in Python?

Ans - Memory management is the process of allocating, using, and releasing memory for objects during program execution. Python manages most object memory automatically. In CPython, reference counting is used along with a cyclic garbage collector, while the interpreter's allocator manages memory blocks.


Q12. What are the basic steps involved in exception handling in Python?

Ans - 1. Put code that may fail inside a try block. 2. Add one or more except blocks to handle specific exceptions. 3. Optionally use else for code that should run only if no exception occurred. 4. Optionally use finally for cleanup that should run either way. 5. If appropriate, raise an exception to report an error that the current function cannot handle.


Q13. Why is memory management important in Python?

Ans - Good memory management helps a program use resources efficiently, remain responsive, and avoid excessive memory consumption. Releasing references to unneeded objects allows memory to be reclaimed when possible. Although Python automates much of this work, keeping unnecessary objects or retaining large collections can still cause high memory use.


Q14. What is the role of try and except in exception handling?

Ans - The try block contains code that might raise an exception. If an exception matching an except clause occurs, Python transfers control to that except block, where the program can handle the problem. If no matching except block exists, the exception propagates to the caller and may terminate the program if it remains unhandled.


Q15. How does Python's garbage collection system work?

Ans - In CPython, reference counting reclaims an object when its reference count reaches zero. A cyclic garbage collector also detects groups of container objects that refer to one another but are otherwise unreachable, allowing many reference cycles to be reclaimed. Other Python implementations may use different memory-management details. Garbage collection does not guarantee that all memory is immediately returned to the operating system.


Q16. What is the purpose of the else block in exception handling?

Ans - The else block runs only when the try block completes without raising an exception. It is useful for keeping success-path code separate from the code being monitored for errors. It does not run if an exception occurs in the try block.


Q17. What are the common logging levels in Python?

Ans - The standard levels, from lowest to highest severity, are DEBUG (detailed diagnostic information), INFO (normal operational events), WARNING (something unexpected or potentially problematic), ERROR (an operation failed), and CRITICAL (a serious failure that may prevent the program from continuing). The logging module also has a NOTSET level used in configuration.


Q18. What is the difference between os.fork() and multiprocessing in Python?

Ans - os.fork() is a Unix/Unix-like operating-system call that creates a child process by duplicating the calling process. It is not available on all operating systems, notably standard Windows Python. The multiprocessing module provides a higher-level, cross-platform API for creating processes, communicating between them, and managing their lifecycle. On some platforms it may use fork, while on others it uses spawn or another start method.


Q19. What is the importance of closing a file in Python?

Ans - Closing a file flushes buffered output and releases the file handle and associated operating-system resources. Leaving files open can waste resources and may delay writing data. Prefer with open(...) as file: because it closes the file automatically when the block exits.


Q20. What is the difference between file.read() and file.readline() in Python?

Ans - file.read(size) reads the requested number of characters/bytes, or the remaining content if size is omitted, and returns it as one string or bytes object. file.readline() reads one line, including its newline character when present, and returns an empty string at end-of-file. For large files, iterating over the file object is usually more memory-efficient than reading everything at once.


Q21. What is the logging module in Python used for?

Ans - The built-in logging module records application events with different severity levels and configurable formats and destinations. It supports loggers, handlers, formatters, and filters, so messages can be routed to the console, files, or other outputs. It is commonly used for debugging, monitoring, and diagnosing production problems.


Q22. What is the os module in Python used for in file handling?

Ans - The os module provides operating-system-related functions, including checking whether paths exist, listing directories, creating or removing directories, renaming files, and working with path information. For example, os.path.exists(path) checks whether a path exists and os.path.isfile(path) checks whether it is a file. The pathlib module is also a modern option for path operations.


Q23. What are the challenges associated with memory management in Python?

Ans - Potential challenges include retaining references to objects that are no longer needed, using large lists or caches, creating reference cycles, and misunderstanding that deleting a variable always returns memory to the operating system immediately. Memory leaks can still occur when references are unintentionally retained. Profiling and careful resource management help identify excessive memory use.


Q24. How do you raise an exception manually in Python?

Ans - Use the raise statement followed by an exception instance or exception class. For example: raise ValueError('Age must be positive'). This is useful when input or program state violates a rule. Inside an except block, raise by itself re-raises the current exception.


Q25. Why is it important to use multithreading in certain applications?

Ans - Multithreading can keep an application responsive and improve throughput when tasks spend time waiting for input/output, such as network requests, file operations, or database calls. While one thread waits, another can make progress. Threads share memory, which can simplify data access, but shared state must be synchronized where necessary. CPU-bound Python work in standard CPython may benefit more from multiprocessing because of the GIL.

```python
#Q1. How can you open a file for writing in Python and write a string to it?
with open("output.txt", "w", encoding="utf-8") as file:
    file.write("Hello, Python!\n")

# "w" creates the file if needed and overwrites existing contents.
```

```python
#Q2. Write a Python program to read the contents of a file and print each line.
try:
    with open("input.txt", "r", encoding="utf-8") as file:
        for line in file:
            print(line, end="")
except FileNotFoundError:
    print("The file was not found.")
```

```python
#Q3. How would you handle a case where the file doesn't exist while trying to open it for reading?
try:
    with open("input.txt", "r", encoding="utf-8") as file:
        contents = file.read()
        print(contents)
except FileNotFoundError:
    print("Error: input.txt does not exist.")
except OSError as error:
    print(f"Could not read the file: {error}")
```

```python
#Q4. Write a Python script that reads from one file and writes its content to another file.
source_path = "source.txt"
destination_path = "copy.txt"

try:
    with open(source_path, "r", encoding="utf-8") as source:
        contents = source.read()
    with open(destination_path, "w", encoding="utf-8") as destination:
        destination.write(contents)
    print("File copied successfully.")
except OSError as error:
    print(f"File operation failed: {error}")
```

```python
#Q5. How would you catch and handle division by zero error in Python?
try:
    numerator = 10
    denominator = 0
    result = numerator / denominator
    print(result)
except ZeroDivisionError:
    print("Error: division by zero is not allowed.")
```

```python
#Q6. Write a Python program that logs an error message to a log file when a division by zero exception occurs.
import logging

logging.basicConfig(
    filename="errors.log",
    level=logging.ERROR,
    format="%(asctime)s %(levelname)s: %(message)s"
)

try:
    result = 10 / 0
except ZeroDivisionError:
    logging.exception("Division by zero occurred")
    print("An error occurred; see errors.log for details.")
```

```python
#Q18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("my_app")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "application.log",
    maxBytes=1_000_000,  # approximately 1 MB
    backupCount=3,
    encoding="utf-8"
)
handler.setFormatter(logging.Formatter(
    "%(asctime)s %(levelname)s: %(message)s"
))
logger.addHandler(handler)

logger.info("Application started.")
logger.error("Example error message.")

# Rotation keeps the current log and up to three backup files.
```

```python
#Q21. Write a Python program that reads a file and prints the number of occurrences of a specific word.
from pathlib import Path

file_path = Path("input.txt")
target_word = "python"

try:
    text = file_path.read_text(encoding="utf-8")
    # Count whole-word matches, ignoring letter case.
    import re
    count = len(re.findall(
        rf"\b{re.escape(target_word)}\b", text, flags=re.IGNORECASE
    ))
    print(f"'{target_word}' occurs {count} time(s).")
except FileNotFoundError:
    print("File not found.")
```

```python
#Q22. How can you check if a file is empty before attempting to read its contents?
from pathlib import Path

path = Path("input.txt")
if not path.is_file():
    print("File does not exist.")
elif path.stat().st_size == 0:
    print("File is empty.")
else:
    print(path.read_text(encoding="utf-8"))

# File size checks can be affected by concurrent changes; handle OSError
# if the file may change or become inaccessible.
```
